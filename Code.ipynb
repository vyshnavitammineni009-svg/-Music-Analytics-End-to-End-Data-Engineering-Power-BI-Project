```python
import pandas as pd

artist_df = pd.read_csv("artist.csv")
albums_df = pd.read_csv("albums (1).csv")
tracks_df = pd.read_csv("tracks.csv")

artist_df.head(), albums_df.head(), tracks_df.head()
```

```python
artist_df.info()
albums_df.info()
tracks_df.info()
```

```python
artist_df.drop_duplicates(inplace=True)
albums_df.drop_duplicates(inplace=True)
tracks_df.drop_duplicates(inplace=True)
```

```python
artist_df.isna().sum()
albums_df.isna().sum()
tracks_df.isna().sum()
```

```python
# Check referential integrity
albums_df['artist_id'].isin(artist_df['id']).value_counts()
albums_df['track_id'].isin(tracks_df['id']).value_counts()
```

```python
artist_album_df = pd.merge(
    albums_df,
    artist_df,
    left_on="artist_id",
    right_on="id",
    how="left"
)
```

```python
fact_tracks_df = pd.merge(
    tracks_df,
    artist_album_df,
    left_on="id",
    right_on="track_id",
    how="left"
)
```

```python
fact_tracks = fact_tracks_df[[
    'track_id',
    'track_name',
    'album_id',
    'artist_id',
    'duration_ms',
    'track_popularity'
]]
```

```python
dim_artist = artist_df[['id', 'name', 'genre_0']]
dim_album = albums_df[['album_id', 'album_name', 'release_date']]
```

```python
fact_tracks.to_csv("fact_tracks.csv", index=False)
dim_artist.to_csv("dim_artist.csv", index=False)
dim_album.to_csv("dim_album.csv", index=False)
```
